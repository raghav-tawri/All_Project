## Section 2: Missing Value Handling

```python
# Q11. Find total number of missing values in each column
df.isnull().sum()
```

```python
# Q12. Fill missing values in Age column with mean age
df_q12 = df.copy()
df_q12['Age'] = df_q12['Age'].fillna(df_q12['Age'].mean())
print('Missing Age values after fill:', df_q12['Age'].isnull().sum())
```

```python
# Q13. Fill missing values in Embarked with mode
df_q13 = df.copy()
df_q13['Embarked'] = df_q13['Embarked'].fillna(df_q13['Embarked'].mode()[0])
print('Missing Embarked values after fill:', df_q13['Embarked'].isnull().sum())
```

```python
# Q14. Fill missing values in Cabin with 'Unknown'
df_q14 = df.copy()
df_q14['Cabin'] = df_q14['Cabin'].fillna('Unknown')
print('Missing Cabin values after fill:', df_q14['Cabin'].isnull().sum())
```

```python
# Q15. Fill missing Cabin values using forward fill (ffill)
df_q15 = df.copy()
df_q15['Cabin'] = df_q15['Cabin'].ffill()
print('Missing Cabin values after ffill:', df_q15['Cabin'].isnull().sum())
```

```python
# Q16. Fill missing Cabin values using backward fill (bfill)
df_q16 = df.copy()
df_q16['Cabin'] = df_q16['Cabin'].bfill()
print('Missing Cabin values after bfill:', df_q16['Cabin'].isnull().sum())
```

```python
# Q17. New DataFrame where all missing values are replaced with 'Not Available'
df_q17 = df.fillna('Not Available')
df_q17.isnull().sum()
```

# Section 10: Using map()

```python
# Q50. Convert: male → M, female → F
df['Sex_Coded'] = df['Sex'].map({'male': 'M', 'female': 'F'})
df[['Sex', 'Sex_Coded']].head()
```

```python
# Q51. Convert: 0 → Did Not Survive, 1 → Survived
df['Survival_Label'] = df['Survived'].map({0: 'Did Not Survive', 1: 'Survived'})
df[['Survived', 'Survival_Label']].head()
```

```python
# Q52. Convert: 1 → First Class, 2 → Second Class, 3 → Third Class
df['Class_Label'] = df['Pclass'].map({1: 'First Class', 2: 'Second Class', 3: 'Third Class'})
df[['Pclass', 'Class_Label']].head()
```

# Section 11: Using apply() with lambda

```python
# Q53. AgeGroup: Child (<18), Adult (18-60), Senior (>60)
df['AgeGroup'] = df['Age'].apply(
    lambda x: 'Child' if x < 18 else ('Senior' if x > 60 else 'Adult') if pd.notnull(x) else np.nan
)
df[['Name', 'Age', 'AgeGroup']].head(10)
```

```python
# Q54. FareCategory: Low (<20), Medium (20-100), High (>100)
df['FareCategory'] = df['Fare'].apply(
    lambda x: 'Low' if x < 20 else ('High' if x > 100 else 'Medium')
)
df[['Name', 'Fare', 'FareCategory']].head()
```

```python
# Q55. SurvivalStatus: Alive (Survived=1), Dead (Survived=0)
df['SurvivalStatus'] = df['Survived'].apply(lambda x: 'Alive' if x == 1 else 'Dead')
df[['Survived', 'SurvivalStatus']].head()
```

```python
# Q56. ClassType: Premium (Pclass=1), Standard otherwise
df['ClassType'] = df['Pclass'].apply(lambda x: 'Premium' if x == 1 else 'Standard')
df[['Pclass', 'ClassType']].head()
```

```python
# Q57. TicketPrefix: first 3 characters of each ticket number
df['TicketPrefix'] = df['Ticket'].apply(lambda x: str(x)[:3])
df[['Ticket', 'TicketPrefix']].head()
```

```python
# Q70. Create a cleaned DataFrame ready for further analysis
df_clean = df.copy()

# Fill missing Age with median
df_clean['Age'] = df_clean['Age'].fillna(df_clean['Age'].median())

# Fill missing Embarked with mode
df_clean['Embarked'] = df_clean['Embarked'].fillna(df_clean['Embarked'].mode()[0])

# Fill missing Cabin with 'Unknown'
df_clean['Cabin'] = df_clean['Cabin'].fillna('Unknown')

# Drop unnecessary columns
df_clean = df_clean.drop(columns=['PassengerId', 'Ticket'])

# Reset index
df_clean = df_clean.reset_index(drop=True)

print('Cleaned DataFrame Shape:', df_clean.shape)
print('Missing Values:\n', df_clean.isnull().sum())
df_clean.head()
```
