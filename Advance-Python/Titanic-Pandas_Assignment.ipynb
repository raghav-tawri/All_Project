## 📌 Section D: Adding New Columns

```python
# Q31. Create column Family_Size = SibSp + Parch + 1
df['Family_Size'] = df['SibSp'] + df['Parch'] + 1
print('Family_Size column added:')
df[['Name', 'SibSp', 'Parch', 'Family_Size']].head(10)
```

```python
# Q32. Create column Fare_with_Tax = Fare * 1.18
df['Fare_with_Tax'] = df['Fare'] * 1.18
print('Fare_with_Tax column added (18% tax):')
df[['Name', 'Fare', 'Fare_with_Tax']].head(10)
```

```python
# Q33. Create column Is_Child = True if Age < 18
df['Is_Child'] = df['Age'] < 18
print('Is_Child column added:')
df[['Name', 'Age', 'Is_Child']].head(10)
```

```python
# Q34. Create column Name_Length = length of passenger's name
df['Name_Length'] = df['Name'].str.len()
print('Name_Length column added:')
df[['Name', 'Name_Length']].head(10)
```

```python
# Q35. Create column Fare_Category
# Fare > 100  → 'Very High'
# Fare > 50   → 'High'
# Otherwise   → 'Low'

def categorize_fare(fare):
    if fare > 100:
        return 'Very High'
    elif fare > 50:
        return 'High'
    else:
        return 'Low'

df['Fare_Category'] = df['Fare'].apply(categorize_fare)
print('Fare_Category column added:')
print(df['Fare_Category'].value_counts())
df[['Name', 'Fare', 'Fare_Category']].head(10)
```

## 📌 Section E: Updating Values

```python
# Q36. Increase all fares by 10%
print('Before:', df['Fare'].head(5).values)
df['Fare'] = df['Fare'] * 1.10
print('After (+10%):', df['Fare'].head(5).values)
```

```python
# Q37. Replace missing values in Embarked with 'Unknown'
print('Missing Embarked before:', df['Embarked'].isnull().sum())
df['Embarked'] = df['Embarked'].fillna('Unknown')
print('Missing Embarked after:', df['Embarked'].isnull().sum())
df['Embarked'].value_counts()
```

```python
# Q39. Set the fare of the first passenger to 999
print('Fare before:', df.loc[0, 'Fare'])
df.loc[0, 'Fare'] = 999
print('Fare after:', df.loc[0, 'Fare'])
```

## 📌 Section G: Handling Missing Values

```python
# Q45. Count missing values in each column
# Reload fresh copy for clean demonstration
df_fresh = pd.read_csv('train.csv')

missing_per_col = df_fresh.isnull().sum()
print('Missing values per column:')
print(missing_per_col[missing_per_col > 0])
```

```python
# Q46. Total number of missing values in the entire dataset
total_missing = df_fresh.isnull().sum().sum()
print(f'Total missing values in entire dataset: {total_missing}')
```

```python
# Q47. Fill missing values in Age with the mean age
mean_age = df_fresh['Age'].mean()
print(f'Mean Age: {mean_age:.2f}')
print(f'Missing Age before: {df_fresh["Age"].isnull().sum()}')
df_fresh['Age'] = df_fresh['Age'].fillna(mean_age)
print(f'Missing Age after:  {df_fresh["Age"].isnull().sum()}')
```

```python
# Q48. Fill missing values in Embarked with the mode
mode_embarked = df_fresh['Embarked'].mode()[0]
print(f'Mode of Embarked: {mode_embarked}')
print(f'Missing Embarked before: {df_fresh["Embarked"].isnull().sum()}')
df_fresh['Embarked'] = df_fresh['Embarked'].fillna(mode_embarked)
print(f'Missing Embarked after:  {df_fresh["Embarked"].isnull().sum()}')
```

```python
# Q49. Drop all rows containing at least one missing value
df_no_missing = df_fresh.dropna()
print(f'Shape before dropna: {df_fresh.shape}')
print(f'Shape after  dropna: {df_no_missing.shape}')
print(f'Rows removed: {df_fresh.shape[0] - df_no_missing.shape[0]}')
```

```python
# Q50. Compare shape before and after handling missing values
df_original = pd.read_csv('train.csv')
df_cleaned  = df_original.copy()

# Fill Age with mean, Embarked with mode, drop rows still having NaN
df_cleaned['Age']      = df_cleaned['Age'].fillna(df_cleaned['Age'].mean())
df_cleaned['Embarked'] = df_cleaned['Embarked'].fillna(df_cleaned['Embarked'].mode()[0])
df_cleaned             = df_cleaned.dropna()

comparison = pd.DataFrame({
    'Dataset'        : ['Original', 'After Handling Missing Values'],
    'Rows'           : [df_original.shape[0], df_cleaned.shape[0]],
    'Columns'        : [df_original.shape[1], df_cleaned.shape[1]],
    'Missing Values' : [df_original.isnull().sum().sum(), df_cleaned.isnull().sum().sum()]
})

print('📊 Shape Comparison Before vs After Handling Missing Values:')
comparison
```
